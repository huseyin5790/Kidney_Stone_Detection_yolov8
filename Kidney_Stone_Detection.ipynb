##test

```python
import cv2
import imutils
from ultralytics import YOLO
from google.colab.patches import cv2_imshow

font = cv2.FONT_HERSHEY_SIMPLEX

img_path = "inference2/299.jpg"
model_path = "runs/detect/kidney_stone_detection2/weights/best.pt"

model = YOLO(model_path)

img= cv2.imread(img_path)
img = imutils.resize(img, width=360)

results = model(img)[0]
# print(results)

threshold = 0.5
for result in results.boxes.data.tolist():
 x1, y1, x2, y2, score, class_id = result
 x1, y1, x2, y2, class_id = int(x1),int(y1),int(x2),int(y2),int(class_id)
# print(result)
 if score > threshold:
  cv2.rectangle(img, (x1,y1), (x2,y2), (0,0,255), 2)

  class_name = results.names[class_id]
  score = score * 100
  yuvarlanmis_score = round(score)
  text = f"{class_name}: %{yuvarlanmis_score}"

  cv2.putText(img, text,(x1-50,y1-10),font,0.5,(0,0,255), 1, cv2.LINE_AA)

cv2_imshow(img)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import ttach as tta
import ultralytics
from ultralytics import YOLO

import warnings
warnings.filterwarnings('ignore')
warnings.simplefilter('ignore')  # Suppress Matplotlib warnings
plt.rcParams["figure.figsize"] = [3.0, 3.0]

from yolo_cam.eigen_cam import EigenCAM
from yolo_cam.utils.image import show_cam_on_image, scale_cam_image

img = cv2.imread('/content/drive/MyDrive/YOLOV8_2/1_object_detection_V2/data/kidney_stone_dataset/images/test/1085.jpg')
img = cv2.resize(img, (640, 640))
rgb_img = img.copy()
img = np.float32(img) / 255

model = YOLO('/content/drive/MyDrive/YOLOV8_2/1_object_detection_V2/runs/detect/kidney_stone_detection2/weights/best.pt')
model = model.cpu()

target_layers =[model.model.model[-2]]
cam = EigenCAM(model, target_layers,task='cls')
grayscale_cam = cam(rgb_img)[0, :, :]
cam_image = show_cam_on_image(img, grayscale_cam, use_rgb=True)
plt.imshow(cam_image)
plt.show()
```
